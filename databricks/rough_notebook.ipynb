```python
%sql
SELECT
    order_item_id,
    order_id,
    product_id,
    quantity,
    unit_price,
    discount,
    gross_amount,
    discount_amount,
    net_sales
FROM dbt_project_2.gold.fact_sales
WHERE net_sales < 0
ORDER BY net_sales;
```
